### 1. Reverse  a string using for loop

```python
def reverse_string(obj:str) -> str:
    """
    reverse given string

    args:
        obj(str): take only string arguments

    returns: 
        reversed string
    """
    if not isinstance(obj, str):
        raise TypeError("Only accept str type args.")
    if len(obj) < 1:
        raise ValueError("String length should be greater than 0")
    rev_str = ""

    for ch in obj:
        rev_str = ch+rev_str
    return rev_str

    
if __name__ == "__main__":
    inp_str = "Python"
    try:
        print(reverse_string(inp_str))
    except Exception as e:
        print(e)
```

## 2. Largest number in  a list

```python
def largest_number(obj:list[int]) -> int: 
    """ 
    find largest number

    args:
        obj(list[int]): take a list which contain integer value

    returns:
        largest integer value from the list
    """
    if not isinstance(obj, list):
        raise TypeError("only accept list")
    if len(obj) == 0:
        raise ValueError("length of list should be greater than 0")
    largest = obj[0]
    for num in obj:
        if not isinstance(num, int):
            raise TypeError("Value inside list should be integer")
        if num > largest:
            largest = num
    return largest

if __name__ == "__main__":

    lst = [1,5,2,3,999,4,90]
    try:
        print(largest_number(lst))

    except Exception as e:
        print(e)
```

## 3. prime number.

```python
def check_prime(obj: int) -> int:
    """
    Check number is prime or not

    args: 
        obj(int): number
    returns: 
        true, false
    """
    if not isinstance(obj, int):
        raise TypeError("Value should be integer.")
    if obj < 2:
        raise ValueError("Value should be greater than 2 to become a prime number.")
    for i in range(2, int(obj**0.5)+1):
        if obj%i == 0:
            return False
    else:
        return True
                
if __name__ == "__main__":
    num = 499
    try:
        print(check_prime(num))
    except Exception as e:
        print(e)
```

## 4. print prime number between range.

```python
def get_prime_range(start:int, end:int) -> list[int]:
    """
    print prime number between range

    arge: 
        start(int): integer number
        end(int): integer number

    returns:
        list of prime number
    """
    prime_list =[]
    if not isinstance(start, int) and not isinstance(end, int):
        raise TypeError(f"{start, end} both are not integer" )
    if start < 2 or end < start:
        raise ValueError(f"end should be greater than start ({start}) and start should be greater than 2")
    for num in range (start, end+1):
        for i in range(2, int(num**0.5)+1):
            if num%i == 0:
                break
        else:
            prime_list.append(num)
    return prime_list

if __name__ == "__main__":
    try:
        start = int(input("Start: "))
        end = int(input("end: "))
        print(get_prime_range(start, end))
    except Exception as e:
        print(e)
```

## 5. find factorial 

```python
def find_factorial(obj: int) -> int:
    """
    calculate factorial

    arge: 
        obj(int): integer number
        

    returns:
        factorial of integer number
    """

    if not isinstance(obj, int):
        raise TypeError("value should be integer.")
    if obj < 1:
        raise ValueError("Value should be greater than 0")
    fact = 1
    for i in range(1, obj+1):
        fact *= i
    return fact

if __name__ == "__main__":
    try:
        print(find_factorial(5))
    except TypeError as e:
        print(e)
    except ValueError as e:
        print(e)
```

```python
def check_paildrome(obj:str) -> bool:
    """ 
    return True or False
    

    args: 
        obj(str): takes string value
        
        


    returns:
        obj(bool): returns true or false.
    """

    if not isinstance(obj, str):
        raise TypeError("value should be str.")
    if len(obj) < 1:
        raise ValueError("length should be greater than 0")
    temp = ""
    for item in obj:
        temp = item+temp
    if temp == obj:
        return True
    else:
        return False

if __name__ == "__main__":
    inp = input("Enter value to check palindrome: ")
    try:
        print(check_paildrome(inp))
    except ValueError as e:
        print("ValueError: ", e)
    except TypeError as e:
        print("TypeError: ",e)    
```

## 9. find sum of digit of a number.

```python
def sum_digit(obj:int)->int:
    """ 
    return sum of all digit.
    

    args: 
        obj(int): takes list value
        
        


    returns:
        obj(int): returns sum of digits.
    """
    if obj == 0:
        return 0
    if not isinstance(obj, int):
        raise TypeError("Value shoould be integer.")
    sum = 0 
    while obj != 0:
        sum = sum + obj%10
        obj = obj // 10
    return abs(sum)
if __name__ == "__main__":
    try:
        print(sum_digit(12345))
    except Exception as e:
        print(e)
```

## 10. remove duplicate

```python
def remove_duplicate(obj:list[int]) -> list[int]:
    """ 
    return list without containing duplicate item
    

    args: 
        obj(list[int]): takes list value
        
        


    returns:
        obj(list[int]): returns list elements.
    """
    if not isinstance(obj, list):
        raise TypeError("accept only list arguments")
    if len(obj) < 1:
        raise ValueError("length of list should be greater than 0.")
    list2 = []
    for item in obj:
        if not isinstance(item, int):
            raise TypeError("all list item should be int")
        if not item in list2:
            list2.append(item)
    return list2

if __name__ == "__main__":
    try:
        print(remove_duplicate([1,2,2,3]))
    except (TypeError, ValueError) as e:
        print(e)
    except Exception as e:
        print(e)
```

## 12. each occurence of character in sentance 

```python
def char_count(obj:str) -> dict:
    """
    remove duplicate from inside a list

    args:
        obj(list): accept list 
 
    returns:
        obj(list): list value
    """
    if not isinstance(obj, str):
        raise TypeError("Value should be integer.")
    if len(obj) < 1:
        raise ValueError("Value should be greater than 0")
    char_counter = {}
    for item in obj.upper().replace(" ",""):
        if item in char_counter:
            char_counter[item] += 1
        else:
            char_counter[item] = 1
    return char_counter

if __name__ == "__main__":
    
    try:
        counts = char_count("My name is Shikhar")
        for item, count in counts.items():
            print(item, count)
    except(ValueError, TypeError) as e:
        print(e)
    except Exception as e:
        print(e)
```

## 13. Check Anagram String

```python
def is_anagram(s1, s2):
    """
    Check whether two strings are anagrams.

    An anagram means both strings contain the same characters
    with the same frequency, possibly in different order.

    Args:
        s1 (str): First input string
        s2 (str): Second input string

    Returns:
        bool: True if strings are anagrams, False otherwise

    Raises:
        TypeError: If inputs are not strings
        ValueError: If inputs are empty
    """
    try:
      
        if not isinstance(s1, str) or not isinstance(s2, str):
            raise TypeError("Both inputs must be strings")

        if not s1 or not s2:
            raise ValueError("Strings cannot be empty")

        s1 = s1.replace(" ", "").lower()
        s2 = s2.replace(" ", "").lower()

        if len(s1) != len(s2):
            return False

        
        count = {}
        for ch in s1:
            count[ch] = count.get(ch, 0) + 1

        for ch in s2:
            if ch not in count:
                return False
            count[ch] -= 1
            if count[ch] == 0:
                del count[ch]

        return len(count) == 0

    except Exception as e:
        print(f"Error: {e}")
        return False
print(is_anagram("Listen", "Silent"))   
print(is_anagram("Hello", "World"))     
print(is_anagram("", "test"))             
print(is_anagram(123, "abc"))             
```

## 14. FIbonaaci sequence in python. 

```python


def fibonacci(n: int) -> list[int]:
    """
    Generate the Fibonacci sequence up to n terms.

    Args:
        n (int): Number of terms in the Fibonacci sequence.

    Returns:
        List[int]: A list containing the Fibonacci sequence.

    Raises:
        TypeError: If n is not an integer.
        ValueError: If n is less than or equal to 0.
    """
    try:
        # Type checking
        if not isinstance(n, int):
            raise TypeError("Input must be an integer.")

        # Value checking
        if n <= 0:
            raise ValueError("Input must be a positive integer.")

        sequence: List[int] = []

        # Generate Fibonacci sequence
        a, b = 0, 1
        for _ in range(n):
            sequence.append(a)
            a, b = b, a + b

        return sequence

    except (TypeError, ValueError) as e:
        print(f"Error: {e}")
        return []


# Example usage
if __name__ == "__main__":
    result = fibonacci(10)
    print("Fibonacci sequence:", result)
```

## 16:Python program to check power of a number without using ** operator

```python
def power(val, itr):
    
    """
    calculate power of given value

    Args: 
        val(int): input integer
        itr(int): input integer

    return:
        obj(int)
    """
    if not isinstance(val, int) or not isinstance(itr, int):
        raise TypeError("Value should be integer.")
   
    res = 1
    for i in range(itr):
        res *= val
    return res

if __name__ == "__main__":
    try:
        print(power(2,5))
    except (TypeError, IndexError) as e:
        print(e)
    except Exception as e:
        print(e)
```

## 18:Binary search in python

```python
def binary_search(obj:(list[int]), search(int)) -> int:
    
    """
    searches elements uding binary search algorithm

    Args: 
        obj(list[int]): input integer inside a list
        search(int): input integer

    return:
        obj(int)
    """
    if not is isinstance(obj, list) or not isinstance(search, int):
        raise TypeError("first arg should be list of integer and second args should be integer")
    if len(obj) < 0:
        raise ValueError("len of list should be grater than 0")
    low = 0
    high = len(obj) - 1

    while low <= high:
        mid = (low + high) // 2

        if obj[mid] == search:
            return mid
        elif search > obj[mid]:
            low = mid + 1
        else:
            high = mid - 1

    return -1  



if __name__ == "__main__":
    try:
        print(binary_search([1,2,3,4,5,6,7,8,9], 5)) 
    except (TypeError, IndexError) as e:
        print(e)
    except Exception as e:
        print(e)
```

## 20:python program to find gcd using Euclidean Algorithm

```python
def calculate_gcd(n1:int, n2:int) -> int:
    """
    find gcd

    Args: 
        n1(int): input integer
        n2(int): input integer

    return:
        gcd(int)
    """
    if not isinstance(n1, int) or not isinstance(n2, int):
        raise TypeError("args shoulod be integer.")
    while n2 != 0:
        n1,n2 = n2, n1%n2
    return n1
    
    

if __name__ == "__main__":
    try:
        print(calculate_gcd(20,15)) 
    except (TypeError, IndexError) as e:
        print(e)
    except Exception as e:
        print(e)
```

## 23. Decimal to binary

```python
def dec_to_bin(obj:int) -> int:
    """ 
    return binary of decimal numbers
    

    args: 
        obj(int): takes int value
        


    returns:
        obj(int): returns integer value(binary number).
    """
    if obj == 1:
        return 1
    if obj == 0:
        return 0
    if obj < 0:
        raise ValueError("negative value is not accepted.")
    if not isinstance(obj, int):
        raise TypeError("Value should be integer.")
    decimal = 0
    while obj != 0:
        rem = obj%2
        decimal = (decimal*10) + rem
        obj //= 2
    return decimal

if __name__ == "__main__":
    try:
        print(dec_to_bin(1))
    except Exception as e:
        print(e)
```

## 24. Remove Whitespace

```python
def remove_whitesapce(obj: str) -> str:
    """ 
    return clean string
    

    args: 
        obj(str): takes str value
        


    returns:
        obj(str): returns clean str value.
    """
    if len(obj) < 1:
        raise ValueError("can not accept empty string")
    
    if not isinstance(obj, str):
        raise TypeError("Value should be str.")
    res = ""
    space_flag = False
    for ch in obj:
        if ch in [" ", "\n", "\t"]:
            if not space_flag:
                res += " "
                space_flag = True
        else:
            res += ch
            space_flag = False

    return res.strip()

if __name__ == "__main__":
    try:
        print(remove_whitesapce("   sj  \n  \t   fkdvuf vdf   "))
    except Exception as e:
        print(e)
```

## 25. intersaction of two list

```python
def list_intersaction(l1: list[int], l2:list[int]) -> list:
    """ 
    return intersaction of two list
    

    args: 
        l1(list[int]): takes list value
        l2(list[int]): takes list value
        


    returns:
        obj(list[int]): returns list intersaction.
    """
    if len(l1) < 1 or len(l2)<1:
        raise ValueError("can not accept empty list")
    
    if not isinstance(l1, list) or not isinstance(l2, list):
        raise TypeError("Value should be list")
    l3 = []
    for l1_item in l1:
        if not l1_item in l2:
            l3.append(l1_item)

    return l3

if __name__ == "__main__":
    try:
        print(list_intersaction([1,2,3],[2,3,4]) )
    except Exception as e:
        print(e)
```

## 26. SUM of List using recurtion

```python
def sum_of_lst(obj:list[int]) -> int:
    """ 
    return Sum of  list
    

    args: 
        obj(list[int]): takes list value
        
        


    returns:
        obj(int): returns Sum of list elements.
    """

    if len(obj) < 1:
        raise ValueError("can not accept empty list")
    
    if not isinstance(obj, list):
        raise TypeError("Value should be list")

    if len(obj) == 0:
        return 0
    return obj[0]+sum_of_lst(obj[1::])


if __name__ == "__main__":
    try:
        print(sum_of_lst([1,2,3,4]))
    except Exception as e:
        print(e)
```
