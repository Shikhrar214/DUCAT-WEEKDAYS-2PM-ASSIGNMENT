```python
def reverse_string(obj:str) -> str:
    """
    reverse given string

    args:
        obj(str): take only string arguments

    returns: 
        reversed string
    """
    if not isinstance(obj, str):
        raise TypeError("Only accept str type args.")
    if len(obj) < 1:
        raise ValueError("String length should be greater than 0")
    rev_str = ""

    for ch in obj:
        rev_str = ch+rev_str
    return rev_str

    
if __name__ == "__main__":
    inp_str = "Python"
    try:
        print(reverse_string(inp_str))
    except Exception as e:
        print(e)
```

## 2. Largest number in  a list

```python
def largest_number(obj:list[int]) -> int: 
    """ 
    find largest number

    args:
        obj(list[int]): take a list which contain integer value

    returns:
        largest integer value from the list
    """
    if not isinstance(obj, list):
        raise TypeError("only accept list")
    if len(obj) == 0:
        raise ValueError("length of list should be greater than 0")
    largest = obj[0]
    for num in obj:
        if not isinstance(num, int):
            raise TypeError("Value inside list should be integer")
        if num > largest:
            largest = num
    return largest

if __name__ == "__main__":

    lst = [1,5,2,3,999,4,90]
    try:
        print(largest_number(lst))

    except Exception as e:
        print(e)
```

## 3. prime number.

```python
def check_prime(obj: int) -> int:
    """
    Check number is prime or not

    args: 
        obj(int): number
    returns: 
        true, false
    """
    if not isinstance(obj, int):
        raise TypeError("Value should be integer.")
    if obj < 2:
        raise ValueError("Value should be greater than 2 to become a prime number.")
    for i in range(2, int(obj**0.5)+1):
        if obj%i == 0:
            return False
    else:
        return True
                
if __name__ == "__main__":
    num = 499
    try:
        print(check_prime(num))
    except Exception as e:
        print(e)
```

## 4. print prime number between range.

```python
def get_prime_range(start:int, end:int) -> list[int]:
    """
    print prime number between range

    arge: 
        start(int): integer number
        end(int): integer number

    returns:
        list of prime number
    """
    prime_list =[]
    if not isinstance(start, int) and not isinstance(end, int):
        raise TypeError(f"{start, end} both are not integer" )
    if start < 2 or end < start:
        raise ValueError(f"end should be greater than start ({start}) and start should be greater than 2")
    for num in range (start, end+1):
        for i in range(2, int(num**0.5)+1):
            if num%i == 0:
                break
        else:
            prime_list.append(num)
    return prime_list

if __name__ == "__main__":
    try:
        start = int(input("Start: "))
        end = int(input("end: "))
        print(get_prime_range(start, end))
    except Exception as e:
        print(e)
```

## 5. find factorial 

```python
def find_factorial(obj: int) -> int:
    """
    calculate factorial

    arge: 
        obj(int): integer number
        

    returns:
        factorial of integer number
    """

    if not isinstance(obj, int):
        raise TypeError("value should be integer.")
    if obj < 1:
        raise ValueError("Value should be greater than 0")
    fact = 1
    for i in range(1, obj+1):
        fact *= i
    return fact

if __name__ == "__main__":
    try:
        print(find_factorial(5))
    except TypeError as e:
        print(e)
    except ValueError as e:
        print(e)
```

## 13. Check Anagram String

```python
def is_anagram(s1, s2):
    """
    Check whether two strings are anagrams.

    An anagram means both strings contain the same characters
    with the same frequency, possibly in different order.

    Args:
        s1 (str): First input string
        s2 (str): Second input string

    Returns:
        bool: True if strings are anagrams, False otherwise

    Raises:
        TypeError: If inputs are not strings
        ValueError: If inputs are empty
    """
    try:
      
        if not isinstance(s1, str) or not isinstance(s2, str):
            raise TypeError("Both inputs must be strings")

        if not s1 or not s2:
            raise ValueError("Strings cannot be empty")

        s1 = s1.replace(" ", "").lower()
        s2 = s2.replace(" ", "").lower()

        if len(s1) != len(s2):
            return False

        
        count = {}
        for ch in s1:
            count[ch] = count.get(ch, 0) + 1

        for ch in s2:
            if ch not in count:
                return False
            count[ch] -= 1
            if count[ch] == 0:
                del count[ch]

        return len(count) == 0

    except Exception as e:
        print(f"Error: {e}")
        return False
print(is_anagram("Listen", "Silent"))   
print(is_anagram("Hello", "World"))     
print(is_anagram("", "test"))             
print(is_anagram(123, "abc"))             
```

## 14. FIbonaaci sequence in python. 

```python


def fibonacci(n: int) -> list[int]:
    """
    Generate the Fibonacci sequence up to n terms.

    Args:
        n (int): Number of terms in the Fibonacci sequence.

    Returns:
        List[int]: A list containing the Fibonacci sequence.

    Raises:
        TypeError: If n is not an integer.
        ValueError: If n is less than or equal to 0.
    """
    try:
       
        if not isinstance(n, int):
            raise TypeError("Input must be an integer.")

        
        if n <= 0:
            raise ValueError("Input must be a positive integer.")

        sequence: List[int] = []

        
        a, b = 0, 1
        for _ in range(n):
            sequence.append(a)
            a, b = b, a + b

        return sequence

    except (TypeError, ValueError) as e:
        print(f"Error: {e}")
        return []



if __name__ == "__main__":
    result = fibonacci(10)
    print("Fibonacci sequence:", result)
```

## 18:Binary search in python

```python
def binary_search(obj:(list[int]), search:int) -> int:
    
    """
    searches elements uding binary search algorithm

    Args: 
        obj(list[int]): input integer inside a list
        search(int): input integer

    return:
        obj(int)
    """
    if not isinstance(obj, list) or not isinstance(search, int):
        raise TypeError("first arg should be list of integer and second args should be integer")
    if len(obj) < 0:
        raise ValueError("len of list should be grater than 0")
    low = 0
    high = len(obj) - 1

    while low <= high:
        mid = (low + high) // 2

        if obj[mid] == search:
            return mid
        elif search > obj[mid]:
            low = mid + 1
        else:
            high = mid - 1

    return -1  



if __name__ == "__main__":
    try:
        print(binary_search([1,2,3,4,5,6,7,8,9], 5)) 
    except (TypeError, IndexError) as e:
        print(e)
    except Exception as e:
        print(e)
```

## 23. Decimal to binary

```python
def dec_to_bin(obj:int) -> int:
    """ 
    return binary of decimal numbers
    

    args: 
        obj(int): takes int value
        


    returns:
        obj(int): returns integer value(binary number).
    """
    if obj == 1:
        return 1
    if obj == 0:
        return 0
    if obj < 0:
        raise ValueError("negative value is not accepted.")
    if not isinstance(obj, int):
        raise TypeError("Value should be integer.")
    decimal = 0
    while obj != 0:
        rem = obj%2
        decimal = (decimal*10) + rem
        obj //= 2
    return decimal

if __name__ == "__main__":
    try:
        print(dec_to_bin(1))
    except Exception as e:
        print(e)
```

## 24. Remove Whitespace

```python
def remove_whitesapce(obj: str) -> str:
    """ 
    return clean string
    

    args: 
        obj(str): takes str value
        


    returns:
        obj(str): returns clean str value.
    """
    if len(obj) < 1:
        raise ValueError("can not accept empty string")
    
    if not isinstance(obj, str):
        raise TypeError("Value should be str.")
    res = ""
    space_flag = False
    for ch in obj:
        if ch in [" ", "\n", "\t"]:
            if not space_flag:
                res += " "
                space_flag = True
        else:
            res += ch
            space_flag = False

    return res.strip()

if __name__ == "__main__":
    try:
        print(remove_whitesapce("   sj  \n  \t   fkdvuf vdf   "))
    except Exception as e:
        print(e)
```

## 27. Bubble Sort

```python
def bubble_sort(obj:list[int])->list[int]:
    """
    it sort list values.

    args:
        obj(list[int]): takes int values inside a list

    return:
        obj(list[int]): returns sorted list
    """

    if not isinstance(obj,list):
        raise TypeError("Obj should be list")
    if len(obj) < 0:
        raise ValueError("length should be gretaer than 0.")
    itr = len(obj)-1

    while itr != 0:
        for i in range(len(obj)-1):
            if not isinstance(i, int):
                raise TypeError("value inside list should be integer.")
            if obj[i]>obj[i+1]:
                temp = obj[i]
                obj[i] = obj[i+1]
                obj[i+1] = temp
        itr -= 1

    return obj

if __name__ == "__main__":
    try:
        print(bubble_sort([5, 3, 8, 4, 2]))
    except Exception as e:
        print(e)
```

## 34. Implement Stack using list in python (Lifo)

```python
class Stack:
    def __init__(self):
        self.stack = []
    def is_empty(self):
        return len(self.stack) == 0
    def push(self, item):
        self.stack.append(item)
        return self.stack

    def pop(self):
        if not self.is_empty():
            self.stack.pop()
            return self.stack
        return "list is empty"
    def peek(self):
        if not self.is_empty():
            return self.stack[-1::]
        return "stack is empty."
    def sequence(self):
        if not self.is_empty():
            self.stack.reverse()
            return self.stack
        return "list is empty"
s1 = Stack()

print(s1.is_empty())

print(s1.push(1))
print(s1.push(2))
print(s1.push(3))
print(s1.push(4))

print(s1.peek())

print(s1.sequence())

print(s1.is_empty())
```

## 37. How to flatten a nested list in python.

```python
def flatten_nested_list(obj):
    l = []
    for ol in obj:
        for i in ol:
            l.append(i)
    return l

print(flatten_nested_list([[1, 2, 3], [4, 5, 6], [7, 8, 9]]))
```

```python
def  flat(obj):
    l = [item for sl in obj for item in sl]
    return l
print(flat([[1, 2, 3], [4, 5, 6], [7, 8, 9]]))
```

## 38. Find pair of given sum in a list.

```python
def find_pair(obj, target):

    pl = []

    for index, num in enumerate(obj):

        i = index + 1

        while i < len(obj):

            if num + obj[i] == target:
                pl.append([num, obj[i]])

            i += 1

    return pl


print(find_pair([2, 7, 11, 15, 3, 6], 9))
```

## 41. find median of list in python.

```python
def median(obj):
    obj.sort()
    s = len(obj)
    if s % 2 == 0:
        res = s//2
        med = (obj[res] + obj[res-1])/2
        return med
    else:
        res = s//2
        return obj[res]
print(median([5,4,2,3,1,3,7]))
```

## 43. Find Lcm of two numbers.

```python
def find_lcm(num1, num2):
    smallest = 0 

    if num1 > num2:
        smallest = num2
    else: 
        smallest = num1
    for i in range(2, smallest+1):
        if num1%i == 0 and num2%i == 0:
            return i
    else:
        return 1

find_lcm(50,10)
    
```

## 44. Convert Binary to decimal.

```python
def bin_to_decimal(obj):

    p = 0
    decimal = 0
    while obj != 0:
        rem = obj%10
        if rem == 1:
            d = 1*(pow(2,p))
            decimal += d
            p += 1
        else:
            d = 0
            p += 1
        obj //= 10
    return decimal
print(bin_to_decimal(101))
```

## 46. Write a function to find common element in three list

```python
def common_ele(l1,l2,l3):
    sorter = l1
    if len(l1) < len(l2) or len(l1) < len(l3):
        sorter = l1
    elif len(l2) < len(l3) or len(l2) < len(l1):
        sorter = l2
    else:
        sorter = l3 
    
    common = []
    for i in sorter:
        if i in l2 and i in l3:
            common.append(i)
            return common


common_ele([1,2,3], [3,4,5], [3,6,7])
```

```python
## 47. Write a function to remove nth occurances in a string
```

```python
def remove_nth(obj,/,*, index=None, key=None):
    original = ""

    if index != None:
        for i in range(len(obj)):
            if i == index:
                continue
            
            ch = obj[i]
            original = original+ch
    
    if key != None:
        for i in obj:
            if i == key:
                continue
            original = original+i
    return original
remove_nth("My name is Shikhar.",key="S")
```

```python
## 48. write a function to return the first non repeated character in  a string.
```

```python
def non_repeated_fixed(obj):
    for i in range(len(obj)):
        is_repeated = False
        for j in range(len(obj)):
            if i == j:
                continue
            if obj[i] == obj[j]:
                is_repeated = True
                break  
        
        if not is_repeated:
            return obj[i]
    return "No non-repeated character found"

print(non_repeated_fixed("ShikharS")) 
```
