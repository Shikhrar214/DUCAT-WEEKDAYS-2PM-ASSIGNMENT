### Reverse  a string using for loop

```python
def reverse_string(obj:str) -> str:
    """
    reverse given string

    args:
        obj(str): take only string arguments

    returns: 
        reversed string
    """
    if not isinstance(obj, str):
        raise TypeError("Only accept str type args.")
    if len(obj) < 1:
        raise ValueError("String length should be greater than 0")
    rev_str = ""

    for ch in obj:
        rev_str = ch+rev_str
    return rev_str

    
if __name__ == "__main__":
    inp_str = "Python"
    try:
        print(reverse_string(inp_str))
    except Exception as e:
        print(e)
```

## Largest number in  a list

```python
def largest_number(obj:list[int]) -> int: 
    """ 
    find largest number

    args:
        obj(list[int]): take a list which contain integer value

    returns:
        largest integer value from the list
    """
    if not isinstance(obj, list):
        raise TypeError("only accept list")
    if len(obj) == 0:
        raise ValueError("length of list should be greater than 0")
    largest = obj[0]
    for num in obj:
        if not isinstance(num, int):
            raise TypeError("Value inside list should be integer")
        if num > largest:
            largest = num
    return largest

if __name__ == "__main__":

    lst = [1,5,2,3,999,4,90]
    try:
        print(largest_number(lst))

    except Exception as e:
        print(e)
```

## prime number.

```python
def check_prime(obj: int) -> int:
    """
    Check number is prime or not

    args: 
        obj(int): number
    returns: 
        true, false
    """
    if not isinstance(obj, int):
        raise TypeError("Value should be integer.")
    if obj < 2:
        raise ValueError("Value should be greater than 2 to become a prime number.")
    for i in range(2, int(obj**0.5)+1):
        if obj%i == 0:
            return False
    else:
        return True
                
if __name__ == "__main__":
    num = 499
    try:
        print(check_prime(num))
    except Exception as e:
        print(e)
```

## print prime number between range.

```python
def get_prime_range(start:int, end:int) -> list[int]:
    """
    print prime number between range

    arge: 
        start(int): integer number
        end(int): integer number

    returns:
        list of prime number
    """
    prime_list =[]
    if not isinstance(start, int) and not isinstance(end, int):
        raise TypeError(f"{start, end} both are not integer" )
    if start < 2 or end < start:
        raise ValueError(f"end should be greater than start ({start}) and start should be greater than 2")
    for num in range (start, end+1):
        for i in range(2, int(num**0.5)+1):
            if num%i == 0:
                break
        else:
            prime_list.append(num)
    return prime_list

if __name__ == "__main__":
    try:
        start = int(input("Start: "))
        end = int(input("end: "))
        print(get_prime_range(start, end))
    except Exception as e:
        print(e)
```

## find factorial 

```python
def find_factorial(obj: int) -> int:

    if not isinstance(obj, int):
        raise TypeError("value should be integer.")
    if obj < 1:
        raise ValueError("Value should be greater than 0")
    fact = 1
    for i in range(1, obj+1):
        fact *= i
    return fact

if __name__ == "__main__":
    try:
        print(find_factorial(5))
    except TypeError as e:
        print(e)
    except ValueError as e:
        print(e)
```

## check paildrome string

```python
def check_paildrome(obj:str) -> bool:

    if not isinstance(obj, str):
        raise TypeError("value should be str.")
    if len(obj) < 1:
        raise ValueError("length should be greater than 0")
    temp = ""
    for item in obj:
        temp = item+temp
    if temp == obj:
        return True
    else:
        return False

if __name__ == "__main__":
    inp = input("Enter value to check palindrome: ")
    try:
        print(check_paildrome(inp))
    except ValueError as e:
        print("ValueError: ", e)
    except TypeError as e:
        print("TypeError: ",e)    
```
