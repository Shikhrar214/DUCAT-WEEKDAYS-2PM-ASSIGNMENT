### Question 1: Write a Python program to print all the numbers from 1 to 10 using a for loop.

```python
for i in range(10):
    print(1+i)
```

### Question 2: Write a Python program to find the sum of all numbers from 1 to 100 using a for loop.

```python
sum = 0
for i in range(100):
    sum += (i+1)

print(sum)
```

### Question 3: Write a Python program to print the multiplication table of a given number using a
for loop.

```python
num = int(input("Enter a number: "))
r = int(input("Enter range: "))
for i in range(r):
    table = num * (i+1)
    print(table)
```

### Question 4: Write a Python program to count the number of even and odd numbers from a
series of numbers using a for loop.

```python
list1 = [1,2,3,4,5]

for item in list1:
    if(item%2 == 0):
        print(item, " is even")
    else:
         print(item, " is odd")
```

### Question 5: Write a Python program to find the factorial of a number using a for loop.

```python
num = int(input("Enter a number: "))
fact = 1
for i in range(num):
    if(num == 0):
        print("factorial = 1")
    elif(i < 0):
        if(num%2 == 0):
            fact = (fact * (1+i))
    else:
        fact = (fact * (1+i))
print(fact)
```

### Question 6: Write a Python program to print the Fibonacci sequence up to a specified number using a for loop.

```python
num = int(input("Enter number: "))
prev1 = 1
prev2 = 0
for i in range(num):
    curr = prev1 + prev2
    prev1 = prev2
    prev2 = curr
    print(curr)
    
```

###  Question 7: Write a Python program to check if a given number is prime or not using a for loop.

```python
num = int(input("Enter number: "))
prime_list = [2,3,4,5,6,7,8,9]

for i in prime_list:
    if(num == i or num <= 1):
        continue
    if(num%i == 0):
        print(num, " is not prime number.")
        break
else:
    print(num, " is prime number")
   
```

### Question 8: Write a Python program to find the largest element in a list using a for loop.

```python
num_list = [1,34,23,3345,67,34]

greatest = 0
for i in num_list:
    if(greatest < i):
        greatest = i
print("greatest number = ",greatest)
```

### Question 9: Write a Python program to reverse a given string using a for loop.

```python
full_name = "Shikhar Srivastav"
reverse_full_name = ""
for i in full_name:
    reverse_full_name = i + reverse_full_name
print(reverse_full_name)
```

## Question 10: Write a Python program to find the common elements between two lists using a for loop.

```python
l1 = [1,2,3]
l2 = [4,5,1]

for l1_item in l1:
    for l2_item in l2:
        if(l1_item == l2_item):
            print(l1_item, " = ", l2_item)
```
