## Q1. Write a program to calculate the electricity bill

```python
total_unit = int(input("Enter Total Used Unit Of Electricity: "))
total_bill = 0
if(total_unit <= 0):
    total_bill = 0
elif(total_unit <= 100):
    total_bill = 0
    total_unit -= 100
elif(total_unit <= 200):
    total_bill = (total_unit-100)*5
    total_unit = 0
else:
    total_bill = ((total_unit-200)*10) + (100*5)
    total_unit = 0

print(f"Total bill you have to pay: {total_bill}")
print(f"Total Remaining Unit: {total_unit}")
```

## Q3. Accept the age of 4 people and display the youngest one?

```python

age1 = int(input("Enter age of person 1: "))
age2 = int(input("Enter age of person 2: "))
age3 = int(input("Enter age of person 3: "))
age4 = int(input("Enter age of person 4: "))


if age1 <= age2 and age1 <= age3 and age1 <= age4:
    print("Person 1 is youngest with age:", age1)

elif age2 <= age1 and age2 <= age3 and age2 <= age4:
    print("Person 2 is youngest with age:", age2)

elif age3 <= age1 and age3 <= age2 and age3 <= age4:
    print("Person 3 is youngest with age:", age3)

else:
    print("Person 4 is youngest with age:", age4)
```

## Q4. A company decided to give bonus to employee according to following criteria:

```python
employee_salary = int(input("Enter Employee Salary: "))
time_period = int(input("Enter Time period of Service in years: "))
net_bonus_amount = 0
if(time_period > 10):
    net_bonus_amount = employee_salary/10
elif(time_period >= 6 and time_period <= 10):
    net_bonus_amount = (employee_salary*2)/25
else:
    net_bonus_amount = employee_salary/20
print(f"Total Bonus = {net_bonus_amount}")
```
